```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path = (r"C:/Users/PRASUN PARDHI/Desktop/texxxxt.txt") 

data = np.loadtxt(r"C:\Users\PRASUN PARDHI\Desktop\texxxxt.txt")

time = data[:, 0]
ecg = data[:, 1]
fs = 1000 
```

```python
plt.figure(figsize=(10, 4))
plt.plot(time, ecg, color='black', linewidth=1)
plt.xlabel('Time (min)')
plt.ylabel('ECG Amplitude (mV)')
plt.title('ECG Signal (Time Domain)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(time, ecg, color='black', linewidth=1)
plt.xlabel('Time (min)')
plt.ylabel('ECG Amplitude (mV)')
plt.title('ECG Signal (Time Domain)')
plt.xlim([0, 0.1]) 
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
N = len(ecg)


ecg_fft = np.fft.fft(ecg)    
freq = np.fft.fftfreq(N, d=1/fs)  


positive_freqs = freq[:N//2]     
fft_magnitude = np.abs(ecg_fft[:N//2]) * 2 / N     
```

```python
plt.figure(figsize=(10, 4))
plt.plot(positive_freqs, fft_magnitude, color='blue')  
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.title('Single-Sided Frequency Spectrum of ECG')
plt.xlim([0, 60])  
plt.grid(True)
plt.tight_layout()
plt.show()
```
